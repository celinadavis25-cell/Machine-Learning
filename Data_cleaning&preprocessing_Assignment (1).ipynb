```python
from google.colab import files
import pandas as pd
import io
uploaded = files.upload() # opens files picker
file_name = list(uploaded.keys())[0]
df = pd.read_csv(io.BytesIO(uploaded[file_name]))
print(df.head(10))
```

```python
df = df.drop_duplicates()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7,4))

plt.boxplot(df["Exam_Score"].dropna())

plt.title("Exam Score Boxplot")
plt.ylabel("Exam Score")

plt.show()
```

There are no outliers i.e there is no value that is much higher or lower than the rest of the data.
