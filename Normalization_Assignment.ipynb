```python
from google.colab import files
import pandas as pd
import io
uploaded = files.upload() # opens file picker
df = pd.read_csv(io.BytesIO(uploaded["student_performance.csv"]))
print(df.head())
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()

df["Atendance_Normalized"] = scaler.fit_transform(df[["Attendance"]])

df.head(20)
```

```python
encoded_df = pd.get_dummies(df, columns=["Department"], dtype=int)

encoded_df.head(20)
```

```python
df.to_csv("student_performance_cleaned.csv", index=False)
print("Cleaned dataset saved successfully.")
```
