```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("student_performance_cleaned.csv")

df.head(10)
```

```python
plt.figure(figsize=(10,6))

sns.countplot(
    data=df,
    x="Department"
)

plt.title("Number of Students in each Department")

plt.xticks(rotation=15)

plt.show()

```

**OBSERVATION**

The chart shows Data Science has more students than the other departments while Software Engineering has the least.

```python
plt.figure(figsize=(8,6))

sns.histplot(
    data=df,
    x="Quiz_Score",
    bins=15
)

plt.title("Quiz Score Distribution")


plt.show()
```

```python
plt.figure(figsize=(8,6))

plt.plot(df["Quiz_Score"].head(10), marker="o")

plt.title("QUIZ SCORES OF THE FIRST 10 STUDENTS")
plt.xlabel("Student")
plt.ylabel("Quiz Score")

plt.grid(True)

plt.show()
```

**OBSERVATION**

This chart shows that student 7 has the highest quiz score of 92 while student 1 has the lowest score of 44.

```python
plt.figure(figsize=(10,6))

plt.bar(
    df["Gender"].value_counts().index,
    df["Gender"].value_counts().values
)

plt.title("Number of Students by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Students")
plt.xticks(rotation=15)
plt.show()
```

**OBSERVATION**

The chart indicates that there are more female students than the male students.

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x="Attendance",
    y="Assignment_Score"
)

plt.title("RELATIONSHIP BETWEEN ATTENDANCE AND ASSIGNMENT SCORE")
plt.xlabel("Attendance")
plt.ylabel("Assignment_Score")
plt.show()
```

## **OBSERVATION**

The scatter plot shows that some students with lower attendance achieved high assignment scores, while some with high attendance had moderate and very low scores. There is also one outlier showing a high assignment score with very low attendance.

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    data=df,
    y="Quiz_Score"
)

plt.title("QUIZ SCORES")

plt.show()
```

**OBSERVATION**

There are no possible outliers and the median score is around 68.

```python
correlation = df.corr(numeric_only=True)

correlation
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="Oranges"
)

plt.title("CORRELATION HEATMAP")

plt.show()
```

**OBSERVATION**

The heatmap shows that age has no relationship with academic performance, also assignment score and exam score has a strong positive correlation.
